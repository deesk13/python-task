```python
def longest_unique(arr):
    n = len(arr)
    res = 0

    for i in range(n):
        seen = set()

        for j in range(i, n):
            if arr[j] in seen:
                break
            else:
                seen.add(arr[j])
                res = max(res, j - i + 1)

    return res


arr = [1, 2, 3, 1, 4, 5]

answer = longest_unique(arr)

print(answer)
```

```python
def max_subarray_sum(arr):
    current = arr[0]
    maximum = arr[0]

    for i in range(1, len(arr)):
        current = max(arr[i], current + arr[i])
        maximum = max(maximum, current)

    return maximum


arr = [-2, 3, -1, 5, -6, 4]
print(max_subarray_sum(arr))
```

```python
def trap_water(height):
    left = 0
    right = len(height) - 1

    left_max = 0
    right_max = 0

    water = 0

    while left < right:

        if height[left] <= height[right]:

            if height[left] >= left_max:
                left_max = height[left]
            else:
                water += left_max - height[left]

            left += 1

        else:

            if height[right] >= right_max:
                right_max = height[right]
            else:
                water += right_max - height[right]

            right -= 1

    return water


height = [0, 1, 0, 2, 1, 0, 1, 3]
print(trap_water(height))
```

```python
def max_performance(scores):
    current = scores[0]
    maximum = scores[0]

    for i in range(1, len(scores)):
        current = max(scores[i], current + scores[i])
        maximum = max(maximum, current)

    return maximum


scores = [-5, 2, 3, -1, 4, -10]
print(max_performance(scores))
```

```python
def max_product(arr):
    current_max = arr[0]
    current_min = arr[0]
    answer = arr[0]

    for i in range(1, len(arr)):

        if arr[i] < 0:
            current_max, current_min = current_min, current_max

        current_max = max(arr[i], current_max * arr[i])
        current_min = min(arr[i], current_min * arr[i])

        answer = max(answer, current_max)

    return answer


arr = [2, 3, -2, 4]
print(max_product(arr))
```

```python
def longest_unique(arr):
    seen = set()
    left = 0
    maximum = 0

    for right in range(len(arr)):

        while arr[right] in seen:
            seen.remove(arr[left])
            left += 1

        seen.add(arr[right])

        maximum = max(maximum, right - left + 1)

    return maximum


arr = [10, 20, 10, 30, 40, 20]
print(longest_unique(arr))
```

```python
def subarray_sum(arr, target):
    prefix_sum = 0
    count = 0

    frequency = {0: 1}

    for num in arr:
        prefix_sum += num

        required = prefix_sum - target

        if required in frequency:
            count += frequency[required]

        frequency[prefix_sum] = frequency.get(prefix_sum, 0) + 1

    return count


arr = [1, 2, 3, -2, 2]
target = 3

print(subarray_sum(arr, target))
```

```python
def group_anagrams(words):
    groups = {}

    for word in words:

        key = ''.join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())


words = ["eat", "tea", "tan", "ate", "nat", "bat"]

print(group_anagrams(words))
```

```python
def longest_consecutive(arr):
    numbers = set(arr)
    maximum = 0

    for num in numbers:

        # Start only if num is the beginning
        if num - 1 not in numbers:

            current = num
            length = 1

            while current + 1 in numbers:
                current += 1
                length += 1

            maximum = max(maximum, length)

    return maximum


arr = [100, 4, 200, 1, 3, 2]

print(longest_consecutive(arr))
```

```python
def merge_intervals(intervals):

    if not intervals:
        return []

    intervals.sort()

    result = [intervals[0]]

    for current in intervals[1:]:

        last = result[-1]

        if current[0] <= last[1]:

            last[1] = max(last[1], current[1])

        else:

            result.append(current)

    return result


intervals = [[1, 3], [2, 6], [8, 10], [9, 12]]

print(merge_intervals(intervals))
```
